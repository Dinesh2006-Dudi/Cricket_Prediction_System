# Module 8: Interactive Cricket Prediction Dashboard
**Purpose:** Launch the Gradio dashboard with match prediction inputs, model performance, team analysis, and player analysis.

**Note:** This notebook rebuilds the model before launching the dashboard because Colab notebooks do not share variables with one another.

## Shared setup — load the original project datasets
This setup is included so this notebook can run independently in a fresh Google Colab session.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

base_url = 'https://raw.githubusercontent.com/Dinesh2006-Dudi/ML-Datasets/main/archive/'
matches = pd.read_csv(base_url + 'matches.csv')
deliveries = pd.read_csv(base_url + 'deliveries.csv')
batting_stats = pd.read_csv(base_url + 'batting_stats.csv')
bowling_stats = pd.read_csv(base_url + 'bowling_stats.csv')
venues = pd.read_csv(base_url + 'venues.csv')
squads = pd.read_csv(base_url + 'squads.csv')
print('Matches:', matches.shape, '| Deliveries:', deliveries.shape)
```

```python
# Prepare datasets using the cleaning steps from the original notebook.
matches = matches.copy()
deliveries = deliveries.copy()

for col in ['first_ings_score', 'first_ings_wkts', 'second_ings_score', 'second_ings_wkts']:
    if col in matches.columns:
        matches[col] = matches[col].fillna(matches[col].median())
for col in ['balls_left', 'wb_runs', 'wb_wickets']:
    if col in matches.columns:
        matches[col] = matches[col].fillna(0)
for col in ['player_of_the_match', 'top_scorer', 'best_bowling']:
    if col in matches.columns:
        matches[col] = matches[col].fillna('Unknown')
if 'match_winner' in matches.columns:
    matches = matches.dropna(subset=['match_winner']).copy()

for col, value in [('wicket_type', 'No Wicket'), ('player_dismissed', 'Not Out'), ('fielder', 'No Fielder')]:
    if col in deliveries.columns:
        deliveries[col] = deliveries[col].fillna(value)
for col in ['wide', 'noballs', 'legbyes', 'byes']:
    if col in deliveries.columns:
        deliveries[col] = deliveries[col].fillna(0)

# Some later feature calculations require these numeric columns.
for col in ['runs_of_bat', 'extras']:
    if col in deliveries.columns:
        deliveries[col] = pd.to_numeric(deliveries[col], errors='coerce').fillna(0)
deliveries['total_runs'] = deliveries['runs_of_bat'] + deliveries['extras']
print('Cleaned matches:', matches.shape, '| Cleaned deliveries:', deliveries.shape)
```

## Build features, train models, and select the best model

```python
# CELL 11 — Build leakage-safe historical features

from collections import defaultdict

# Map 'balls' to 'deliveries' from the loaded dataset
balls = deliveries

# Ensure correct chronological order.
matches = matches.sort_values(["date", "match_id"]).reset_index(drop=True)

# Match-level innings summary: group first by match and team to sum up across all innings (including super overs)
innings_summary = balls.groupby(
    ["match_no", "batting_team"]
).agg(
    team_runs=("total_runs", "sum"),
    deliveries=("runs_of_bat", "count"),
    wickets=("player_dismissed", lambda x: (x != "Not Out").sum())
).reset_index()

match_innings = {}
for mid, g in innings_summary.groupby("match_no"):
    match_innings[mid] = g.set_index("batting_team").to_dict("index")

# Players who appeared for each team in each match
players_by_match_team = {}

for mid, g in balls.groupby("match_no"):
    players_by_match_team[mid] = {}

    for team_name, tg in g.groupby("batting_team"):
        batters = set(tg["striker"]) | set(tg["player_dismissed"]) # using striker/non-striker concepts
        # In your dataset, the column is 'striker' rather than 'batsman'
        players_by_match_team[mid].setdefault(team_name, {})["batters"] = batters

    for team_name, tg in g.groupby("bowling_team"):
        bowlers = set(tg["bowler"])
        players_by_match_team[mid].setdefault(team_name, {})["bowlers"] = bowlers

# Historical team statistics
team_hist = defaultdict(lambda: {
    "matches": 0,
    "wins": 0,
    "runs_for": 0,
    "runs_against": 0,
    "wickets_for": 0,
    "wickets_against": 0
})

# Historical player batting statistics
bat_hist = defaultdict(lambda: {
    "runs": 0,
    "balls": 0,
    "outs": 0,
    "matches": 0
})

# Historical player bowling statistics
bowl_hist = defaultdict(lambda: {
    "runs_conceded": 0,
    "legal_balls": 0,
    "wickets": 0,
    "matches": 0
})

def get_team_features(team_name):
    s = team_hist[team_name]
    n = s["matches"]

    return {
        "win_rate": s["wins"] / n if n else 0.5,
        "avg_runs_for": s["runs_for"] / n if n else 0,
        "avg_runs_against": s["runs_against"] / n if n else 0,
        "avg_wickets_for": s["wickets_for"] / n if n else 0,
        "avg_wickets_against": s["wickets_against"] / n if n else 0
    }

def get_player_features(team_name, match_id):
    current = players_by_match_team.get(match_id, {}).get(team_name, {})

    batters = current.get("batters", set())
    bowlers = current.get("bowlers", set())

    batting_values = []
    bowling_values = []

    for player in batters:
        s = bat_hist[player]
        if s["matches"] > 0:
            avg_runs = s["runs"] / s["matches"]
            sr = (s["runs"] / s["balls"] * 100) if s["balls"] else 0
            batting_values.append((avg_runs, sr))

    for player in bowlers:
        s = bowl_hist[player]
        if s["matches"] > 0:
            wickets_per_match = s["wickets"] / s["matches"]
            economy = (
                s["runs_conceded"] / (s["legal_balls"] / 6)
                if s["legal_balls"] else 99
            )
            bowling_values.append((wickets_per_match, economy))

    batter_avg_runs = np.mean([x[0] for x in batting_values]) if batting_values else 0
    batter_sr = np.mean([x[1] for x in batting_values]) if batting_values else 0
    bowler_avg_wickets = np.mean([x[0] for x in bowling_values]) if bowling_values else 0
    bowler_economy = np.mean([x[1] for x in bowling_values]) if bowling_values else 0

    return batter_avg_runs, batter_sr, bowler_avg_wickets, bowler_economy


feature_rows = []

for _, m in matches.iterrows():

    match_id = int(m["match_id"])
    team1 = m["team1"]
    team2 = m["team2"]

    t1 = get_team_features(team1)
    t2 = get_team_features(team2)

    p1 = get_player_features(team1, match_id)
    p2 = get_player_features(team2, match_id)

    feature_rows.append({
        "match_id": match_id,
        "stage": m["stage"],
        "venue": m["venue"],
        "team1": team1,
        "team2": team2,
        "toss_winner": m["toss_winner"],
        "toss_decision": m["toss_decision"],

        "team1_win_rate": t1["win_rate"],
        "team2_win_rate": t2["win_rate"],
        "team1_avg_runs_for": t1["avg_runs_for"],
        "team2_avg_runs_for": t2["avg_runs_for"],
        "team1_avg_runs_against": t1["avg_runs_against"],
        "team2_avg_runs_against": t2["avg_runs_against"],
        "team1_avg_wickets_for": t1["avg_wickets_for"],
        "team2_avg_wickets_for": t2["avg_wickets_for"],
        "team1_avg_wickets_against": t1["avg_wickets_against"],
        "team2_avg_wickets_against": t2["avg_wickets_against"],

        "team1_batter_avg_runs": p1[0],
        "team2_batter_avg_runs": p2[0],
        "team1_batter_sr": p1[1],
        "team2_batter_sr": p2[1],
        "team1_bowler_avg_wickets": p1[2],
        "team2_bowler_avg_wickets": p2[2],
        "team1_bowler_economy": p1[3],
        "team2_bowler_economy": p2[3],

        # Target: 1 means team1 won, 0 means team2 won
        "target": int(m["match_winner"] == team1)
    })

    # ---------------- UPDATE HISTORY AFTER FEATURE CREATION ----------------
    innings = match_innings.get(match_id, {})

    for team_name in [team1, team2]:
        opponent = team2 if team_name == team1 else team1

        own = innings.get(team_name, {})
        opp = innings.get(opponent, {})

        s = team_hist[team_name]
        s["matches"] += 1
        s["wins"] += int(m["match_winner"] == team_name)
        s["runs_for"] += own.get("team_runs", 0)
        s["runs_against"] += opp.get("team_runs", 0)
        s["wickets_for"] += opp.get("wickets", 0)
        s["wickets_against"] += own.get("wickets", 0)

    current_balls = balls[balls["match_no"] == match_id]

    for player, pg in current_balls.groupby("striker"):
        s = bat_hist[player]
        s["runs"] += pg["runs_of_bat"].sum()
        s["balls"] += (pg["wide"] == 0).sum()
        s["outs"] += (pg["player_dismissed"] == player).sum()
        s["matches"] += 1

    for player, pg in current_balls.groupby("bowler"):
        s = bowl_hist[player]

        # Byes and leg-byes are not charged to the bowler.
        runs_conceded = (
            pg["total_runs"]
            - pg["byes"]
            - pg["legbyes"]
        ).sum()

        legal_balls = (
            (pg["wide"] == 0) &
            (pg["noballs"] == 0)
        ).sum()

        wickets = pg["wicket_type"].isin(
            ["bowled", "caught", "lbw", "stumped", "caught and bowled"]
        ).sum()

        s["runs_conceded"] += runs_conceded
        s["legal_balls"] += legal_balls
        s["wickets"] += wickets
        s["matches"] += 1

model_df = pd.DataFrame(feature_rows)

print("Final modeling dataset:", model_df.shape)
display(model_df.head())
```

```python
# CELL 12 — Train/test split: 80% training, 20% testing

from sklearn.model_selection import train_test_split

X = model_df.drop(columns=["target", "match_id"])
y = model_df["target"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
print("Training percentage:", len(X_train) / len(model_df) * 100)
print("Testing percentage:", len(X_test) / len(model_df) * 100)
```

```python
# CELL 15 — Train different ML classification models

from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.svm import SVC

# Identify column types dynamically from X_train
categorical_cols = X_train.select_dtypes(include=["object", "category"]).columns.tolist()
numerical_cols = X_train.select_dtypes(include=["int64", "float64"]).columns.tolist()

# Define the preprocessor
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_cols),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols)
    ]
)

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=5,
        min_samples_leaf=5,
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=400,
        max_depth=8,
        min_samples_leaf=3,
        class_weight="balanced",
        random_state=42
    ),

    "XGBoost": XGBClassifier(
        n_estimators=300,
        max_depth=3,
        learning_rate=0.04,
        subsample=0.85,
        colsample_bytree=0.85,
        eval_metric="logloss",
        random_state=42
    ),
    "SVM": SVC(
    kernel="rbf",
    C=1.0,
    gamma="scale",
    class_weight="balanced",
    random_state=42
),
}

trained_models = {}
results = []

for name, model in models.items():

    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    pipeline.fit(X_train, y_train)

    y_pred = pipeline.predict(X_test)

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred, zero_division=0),
        "Recall": recall_score(y_test, y_pred, zero_division=0),
        "F1-Score": f1_score(y_test, y_pred, zero_division=0)
    })

    trained_models[name] = pipeline

results_df = pd.DataFrame(results).sort_values(
    by="F1-Score",
    ascending=False
).reset_index(drop=True)

display(results_df)
```

```python
# CELL 17 — Automatically select the two best models

top_two = results_df.head(2)

print("Two best models based on F1-score:")
display(top_two)

best_model_name = top_two.iloc[0]["Model"]
second_best_model_name = top_two.iloc[1]["Model"]

best_model = trained_models[best_model_name]
second_best_model = trained_models[second_best_model_name]

print("BEST MODEL:", best_model_name)
print("SECOND BEST MODEL:", second_best_model_name)
```

```python
import gradio as gr
import plotly.express as px
import pandas as pd
import numpy as np

# Correcting variable populations using accurate column names
teams = sorted(set(matches["team1"]) | set(matches["team2"]))
cities = sorted(venues["city"].dropna().unique().tolist())
venues_list = sorted(matches["venue"].dropna().unique().tolist())
seasons = sorted(matches["season"].dropna().astype(int).unique().tolist())

# Correct 'winner' to 'match_winner'
team_wins = matches["match_winner"].value_counts().reset_index()
team_wins.columns = ["Team", "Wins"]
wins_fig = px.bar(team_wins, x="Team", y="Wins", title="Historical Wins by Team")

# Correct 'batsman' -> 'striker', 'batsman_runs' -> 'runs_of_bat'
bat_top = (
    deliveries.groupby("striker")["runs_of_bat"]
    .sum().nlargest(10).sort_values().reset_index()
)
bat_top.columns = ["Player", "Runs"]
bat_fig = px.bar(bat_top, x="Runs", y="Player", orientation="h", title="Top 10 Batsmen")

# Filter deliveries by standard bowler wickets
bower_wicket_kinds = ["bowled", "caught", "lbw", "stumped", "caught and bowled"]
bowl_top = (
    deliveries[deliveries["wicket_type"].str.strip().str.lower().isin(bower_wicket_kinds)]
    .groupby("bowler").size().nlargest(10).sort_values().reset_index(name="Wickets")
)
bowl_fig = px.bar(bowl_top, x="Wickets", y="bowler", orientation="h", title="Top 10 Bowlers")

model_fig = px.bar(
    results_df,
    x="Model",
    y=["Accuracy", "Precision", "Recall", "F1-Score"],
    barmode="group",
    range_y=[0, 1],
    title="Model Performance"
)
model_fig.update_yaxes(tickformat=".0%")

# Helper function to get players for a team
def get_players_for_team(team_name):
    if "squads" in globals() or "squads" in locals():
        # Safe casting with astype(str) to avoid 'str' object has no attribute 'str' error
        team_players = squads[squads["team_name"].astype(str).str.strip().str.lower() == str(team_name).strip().lower()]["player"].dropna().unique().tolist()
        if team_players:
            return sorted(team_players)
    # Fallback to deliveries players if squad sheet is empty or has mismatched names
    team_players = sorted(set(deliveries[deliveries["batting_team"] == team_name]["striker"]) | set(deliveries[deliveries["bowling_team"] == team_name]["bowler"]))
    return team_players if team_players else ["Default Player"]

# Dynamic dropdown updater
def update_team_players(team1, team2):
    p1_choices = get_players_for_team(team1)
    p2_choices = get_players_for_team(team2)
    return (
        gr.Dropdown(choices=p1_choices, value=p1_choices[0] if p1_choices else None),
        gr.Dropdown(choices=p2_choices, value=p2_choices[0] if p2_choices else None)
    )

# Updated predict_dashboard function to align probability and winner prediction correctly
def predict_dashboard(team1, team2, toss_winner, toss_decision, city, venue, p1, p2):
    t1_feats = get_team_features(team1)
    t2_feats = get_team_features(team2)

    p1_feats = (25.0, 140.0, 0.8, 9.0)
    p2_feats = (25.0, 140.0, 0.8, 9.0)

    input_data = pd.DataFrame([{
        "stage": "League",
        "venue": venue,
        "team1": team1,
        "team2": team2,
        "toss_winner": toss_winner,
        "toss_decision": toss_decision,
        "team1_win_rate": t1_feats["win_rate"],
        "team2_win_rate": t2_feats["win_rate"],
        "team1_avg_runs_for": t1_feats["avg_runs_for"],
        "team2_avg_runs_for": t2_feats["avg_runs_for"],
        "team1_avg_runs_against": t1_feats["avg_runs_against"],
        "team2_avg_runs_against": t2_feats["avg_runs_against"],
        "team1_avg_wickets_for": t1_feats["avg_wickets_for"],
        "team2_avg_wickets_for": t2_feats["avg_wickets_for"],
        "team1_avg_wickets_against": t1_feats["avg_wickets_against"],
        "team2_avg_wickets_against": t2_feats["avg_wickets_against"],
        "team1_batter_avg_runs": p1_feats[0],
        "team2_batter_avg_runs": p2_feats[0],
        "team1_batter_sr": p1_feats[1],
        "team2_batter_sr": p2_feats[1],
        "team1_bowler_avg_wickets": p1_feats[2],
        "team2_bowler_avg_wickets": p2_feats[2],
        "team1_bowler_economy": p1_feats[3],
        "team2_bowler_economy": p2_feats[3]
    }])

    # Predict class (1 = Team 1 wins, 0 = Team 2 wins)
    pred_class = int(best_model.predict(input_data)[0])

    try:
        # Ensure class mapping index (0 and 1) corresponds exactly to the models probability outputs
        probs = best_model.predict_proba(input_data)[0]
        # Mapping index 0 -> Team 2, index 1 -> Team 1
        prob_team2 = float(probs[0])
        prob_team1 = float(probs[1])
    except Exception:
        prob_team1 = 0.6 if pred_class == 1 else 0.4
        prob_team2 = 1.0 - prob_team1

    # Ensure predicted winner matches the highest probability mathematically
    if prob_team1 >= prob_team2:
        winner = team1
        highest_prob = prob_team1
    else:
        winner = team2
        highest_prob = prob_team2

    res_md = f"## 🔮 Predicted Winner: **{winner}**"

    # Probability chart
    prob_df = pd.DataFrame({"Team": [team1, team2], "Probability": [prob_team1, prob_team2]})
    fig = px.bar(prob_df, x="Probability", y="Team", orientation="h", title="Win Probability", range_x=[0, 1])
    fig.update_layout(height=250)

    # Team comparison df
    comparison_df = pd.DataFrame({
        "Metric": ["Team 1 Name", "Team 2 Name", "Toss Winner", "Toss Decision", "Team 1 Star Player", "Team 2 Star Player"],
        "Value": [team1, team2, toss_winner, toss_decision, p1, p2]
    })

    explanation = f"### Explanation\nThe machine learning model predicts **{winner}** has a higher chance of winning ({highest_prob:.1%}) based on historically analyzed player statistics, venue advantages, and toss decisions."
    return res_md, fig, comparison_df, explanation

def reset_dashboard_inputs():
    t1 = teams[0] if teams else None
    t2 = teams[1] if len(teams) > 1 else None
    p1_opts = get_players_for_team(t1) if t1 else []
    p2_opts = get_players_for_team(t2) if t2 else []
    return (
        t1,
        t2,
        seasons[-1] if seasons else None,
        t1,
        "bat",
        cities[0] if cities else None,
        venues_list[0] if venues_list else None,
        p1_opts[0] if p1_opts else None,
        p2_opts[0] if p2_opts else None,
        "",
        None,
        None,
        ""
    )

# =========================
# STEP 27 - GRADIO UI DASHBOARD
# =========================
with gr.Blocks(title="Cricket Prediction System") as dashboard:
    gr.Markdown("""
    # 🏏 Cricket Prediction System
    ### Machine Learning Based Match Winner Prediction
    """)

    with gr.Tab("🏆 Match Prediction"):
        with gr.Row():
            with gr.Column():
                team1_input = gr.Dropdown(choices=teams, label="Team 1", value=teams[0] if teams else None)
                team2_input = gr.Dropdown(choices=teams, label="Team 2", value=teams[1] if len(teams) > 1 else None)
                season_input = gr.Dropdown(choices=seasons, label="Season", value=seasons[-1] if seasons else None)
                toss_winner_input = gr.Dropdown(choices=teams, label="Toss Winner", value=teams[0] if teams else None)
                toss_decision_input = gr.Radio(choices=["bat", "field"], label="Toss Decision", value="bat")
            with gr.Column():
                city_input = gr.Dropdown(choices=cities, label="City", value=cities[0] if cities else None)
                venue_input = gr.Dropdown(choices=venues_list, label="Venue", value=venues_list[0] if venues_list else None)

                t1_init_players = get_players_for_team(teams[0]) if teams else []
                t2_init_players = get_players_for_team(teams[1]) if len(teams) > 1 else []

                p1_input = gr.Dropdown(choices=t1_init_players, label="Team 1 Key Player", value=t1_init_players[0] if t1_init_players else None)
                p2_input = gr.Dropdown(choices=t2_init_players, label="Team 2 Key Player", value=t2_init_players[0] if t2_init_players else None)

        # Update player choices dynamically when teams are altered
        team1_input.change(fn=update_team_players, inputs=[team1_input, team2_input], outputs=[p1_input, p2_input])
        team2_input.change(fn=update_team_players, inputs=[team1_input, team2_input], outputs=[p1_input, p2_input])

        with gr.Row():
            predict_button = gr.Button("🔮 Predict Match Winner", variant="primary")
            reset_button = gr.Button("🔄 Reset Dashboard Inputs", variant="secondary")

        result_output = gr.Markdown()
        probability_output = gr.Plot()
        comparison_output = gr.Dataframe(interactive=False)
        explanation_output = gr.Markdown()

        predict_button.click(
            fn=predict_dashboard,
            inputs=[
                team1_input, team2_input, toss_winner_input,
                toss_decision_input, city_input, venue_input,
                p1_input, p2_input
            ],
            outputs=[
                result_output, probability_output,
                comparison_output, explanation_output
            ]
        )

        reset_button.click(
            fn=reset_dashboard_inputs,
            inputs=[],
            outputs=[
                team1_input, team2_input, season_input, toss_winner_input,
                toss_decision_input, city_input, venue_input, p1_input, p2_input,
                result_output, probability_output, comparison_output, explanation_output
            ]
        )

    with gr.Tab("📊 Model Performance"):
        gr.Markdown(f"## Best Model: **{best_model_name}**")
        gr.Dataframe(results_df, interactive=False)
        gr.Plot(model_fig)

    with gr.Tab("🏆 Team Analysis"):
        gr.Plot(wins_fig)

    with gr.Tab("👤 Player Analysis"):
        gr.Plot(bat_fig)
        gr.Plot(bowl_fig)

print("Dashboard is ready.")
dashboard.launch(share=True)
```
