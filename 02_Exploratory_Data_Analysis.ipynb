                                Exploratory Data Analysis

```python
import pandas as pd
import matplotlib.pyplot as plt

# Ensure datasets are loaded into the local workspace if not already present
base_url = "https://raw.githubusercontent.com/Dinesh2006-Dudi/ML-Datasets/main/archive/"
if 'matches' not in locals() and 'matches' not in globals():
    matches = pd.read_csv(base_url + "matches.csv")
if 'deliveries' not in locals() and 'deliveries' not in globals():
    deliveries = pd.read_csv(base_url + "deliveries.csv")

# Point 'balls' to the deliveries DataFrame
balls = deliveries

plt.figure(figsize=(12, 8))

# 1. Winner distribution (using 'match_winner' instead of 'winner')
plt.subplot(2, 2, 1)
matches["match_winner"].value_counts().head(10).plot(kind="bar")
plt.title("Winner Distribution")
plt.xticks(rotation=45, ha='right')

# 2. Season distribution (extracting year from the 'date' column)
plt.subplot(2, 2, 2)
matches["season"] = pd.to_datetime(matches["date"], errors="coerce").dt.year
matches["season"].value_counts().sort_index().plot(kind="bar")
plt.title("Matches by Season")

# 3. Toss decision distribution
plt.subplot(2, 2, 3)
matches["toss_decision"].value_counts().plot(kind="bar")
plt.title("Toss Decision Distribution")

# 4. Runs distribution (using 'runs_of_bat' + 'extras' to get total runs)
plt.subplot(2, 2, 4)
balls["total_runs"] = balls["runs_of_bat"] + balls["extras"]
balls["total_runs"].value_counts().sort_index().plot(kind="bar")
plt.title("Runs per Ball Distribution")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(15, 10))

# 1. Top 10 teams by matches played
plt.subplot(2, 3, 1)
team_count = matches["team1"].value_counts().head(10)
sns.barplot(x=team_count.values, y=team_count.index)
plt.title("Top 10 Teams by Matches")
plt.xlabel("Matches")
plt.ylabel("Team")

# 2. Toss decision distribution
plt.subplot(2, 3, 2)
sns.countplot(data=matches, x="toss_decision")
plt.title("Toss Decision Distribution")
plt.xlabel("Toss Decision")
plt.ylabel("Count")

# 3. Top 10 teams by wins
plt.subplot(2, 3, 3)
wins = matches["match_winner"].value_counts().head(10)
sns.barplot(x=wins.values, y=wins.index)
plt.title("Top 10 Teams by Wins")
plt.xlabel("Wins")
plt.ylabel("Team")

# 4. Winning runs distribution
plt.subplot(2, 3, 4)
sns.histplot(matches["wb_runs"], bins=20, kde=True)
plt.title("Winning Runs Distribution")
plt.xlabel("Winning Runs")
plt.ylabel("Frequency")

# 5. Winning wickets distribution
plt.subplot(2, 3, 5)
sns.histplot(matches["wb_wickets"], bins=10, kde=True)
plt.title("Winning Wickets Distribution")
plt.xlabel("Winning Wickets")
plt.ylabel("Frequency")



plt.tight_layout()
plt.show()
```
